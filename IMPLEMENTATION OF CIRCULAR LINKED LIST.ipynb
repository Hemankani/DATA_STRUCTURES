CREATION

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = Node(None)
        self.tail = Node(None)
        self.head.next = self.tail
        self.tail.next = self.head

    def add(self, data):
        newNode = Node(data)
        if self.head.data is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head.data is None:
            print("List is empty, cannot delete.")
            return

        if self.head == self.tail:
            self.head = Node(None)
            self.tail = Node(None)
            self.head.next = self.tail
            self.tail.next = self.head
            return

        current = self.head
        while current.next != self.tail:
            current = current.next

        current.next = self.head
        self.tail = current
    def display(self):
        current = self.head

        if self.head.data is None and self.head == self.tail:
            print("List is empty")
            return
        else:
            print("Nodes of the circular linked list: ", end="")
            print(current.data, end=" ")
            current = current.next
            while current != self.head:
                print(current.data, end=" ")
                current = current.next
            print()

cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)
cl.display()
```

INSERTION

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        if self.last is not None:
            return self.last
        temp = Node(data)
        self.last = temp
        self.last.next = self.last
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)
        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)
        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        self.last = temp
        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            print("List is empty, cannot add after an item")
            return None

        p = self.last.next
        found_item = False
        while True:
            if p.data == item:
                temp = Node(data)
                temp.next = p.next
                p.next = temp
                if p == self.last:
                    self.last = temp
                found_item = True
                break
            p = p.next
            if p == self.last.next:
                break

        if not found_item:
            print(item, "not present in the list")
        return self.last

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.last.next:
                break
        print()

if __name__ == '__main__':
    llist = CircularLinkedList()

    last = llist.addToEmpty(6)
    last = llist.addBegin(4)
    last = llist.addBegin(2)
    last = llist.addEnd(8)

    last = llist.addEnd(12)
    last = llist.addAfter(10, 8)

    llist.traverse()
```

DELETION

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head is None:
            return

        if self.head == self.tail:
            self.head = None
            self.tail = None
            return

        current = self.head
        while current.next != self.tail:
            current = current.next

        current.next = self.head
        self.tail = current
    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        while True:
            print(current.data, end=" ")
            current = current.next
            if current == self.head:
                break
        print()
cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

print("Original List:")
cl.display()

while cl.head is not None:
    cl.deleteEnd()
    print("Updated List:")
    cl.display()
```

TRAVERSAL

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def push(self, data):
        ptr1 = Node(data)
        ptr1.next = self.head

        if self.head is not None:
            temp = self.head
            while temp.next != self.head:
                temp = temp.next
            temp.next = ptr1
        else:
            ptr1.next = ptr1

        self.head = ptr1

    def printList(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.head:
                break
        print()

cllist = CircularLinkedList()
cllist.push(12)
cllist.push(56)
cllist.push(2)
cllist.push(11)
print("Contents of circular Linked List")
cllist.printList()
```
